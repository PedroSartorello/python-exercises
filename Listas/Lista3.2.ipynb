5. Tuplas como coordenadas:
Receba uma lista de tuplas (x, y) e calcule a distância euclidiana de cada ponto
até a origem.

```python
coordenadas = [(1,2), (3,4), (5,6), (1,1)]

for pontos in coordenadas:
    dist = (pontos[0] ** 2 + pontos[1]**2) ** 0.5
    print(dist)
```

6. Frequência em tupla:
Dada uma tupla de números, retorne um dicionário com a frequência de cada
valor.

```python
tupla = (1,2,3,3,4,5,5,6,7,7,7,7,8,9,10)

count = {}

for item in tupla:
    count[item] = count.get(item, 0) + 1

count
```

7. Ordenar lista de tuplas:
Dada uma lista de tuplas (nome, idade), ordene por idade sem usar sorted
(implemente um algoritmo simples como bubble sort).

```python
pessoas = [
    ("Pedro", 24),
    ("Maria", 19),
    ("João", 31),
    ("Lucas", 24),
    ("Fernanda", 27)
]

for step in range(len(pessoas)-1):
    for i in range(len(pessoas)-1-step):
        if pessoas[i][1] > pessoas[i+1][1]:
            pessoas[i], pessoas[i+1] = pessoas[i+1], pessoas[i]
pessoas
```

9. Interseção de tuplas:
Receba duas tuplas e retorne uma nova tupla com os elementos comuns, sem
duplicatas e na ordem em que aparecem na primeira tupla.

```python
tupla1 = (1, 2, 3, 4, 5, 2, 6, 3)
tupla2 = (3, 5, 7, 2, 8, 3)

conjunto = set(tupla2)
res = []
for num in tupla1:
    if num in conjunto and num not in res:
        res.append(num)

res = tuple(res)
res
```
