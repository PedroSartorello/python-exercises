2. Trabalhando com Listas
Objetivo: Introduzir listas, uma estrutura fundamental para manipulação de dados em Python.
● Crie uma lista com 5 números inteiros.
● Peça ao usuário para adicionar mais um número à lista.
● Remova o menor número da lista.
● Calcule a média dos números restantes.
● Mostre a lista final, o tamanho da lista e a média.

```python
def menorNumero(list):
    menor = list[0]
    idx = 0
    for i, num in enumerate(list):
        if num <= menor:
            menor = num
            idx = i
    return idx

list = [10, 2, 1, 4, 5]
print(list)
number = int(input('Adicione um novo numero'))
list.append(number)
print(list)
list.pop(menorNumero(list))
print(list)
print(len(list))
listMed = sum(list)/len(list)
print(listMed)
```

3. Condicionais e Estruturas de Decisão
Objetivo: Ensinar o uso de condicionais para tomada de decisões.
Crie um programa que solicite ao usuário a nota de um aluno (de 0 a 10).
● Use condicionais para classificar a nota:
● Se a nota for maior ou igual a 7, exiba "Aprovado".
● Se a nota for entre 5 e 6.9, exiba "Recuperação".
● Se a nota for menor que 5, exiba "Reprovado".

```python
nota = float(input("Qual sua nota? "))

while True:
    if nota < 0 or nota > 10:
        print("Nota não válida")
        break

    if nota >= 7:
        print("Aprovado")

    elif nota >= 5 and nota < 7:
        print("Recuperação")

    else:
        print("Reprovado")

    break
```

Objetivo: Revisar todos os conceitos aprendidos em um exercício integrado.
Crie um programa que:
● Solicite ao usuário uma lista de números que cresce enquanto o usuário não digitar
‘sair’.
● Calcule a média, o maior e o menor valor da lista.
● Imprima os valores de maneira amigável (ex: “O maior valor da lista é X”)
● Salve os resultados em um arquivo .txt.

```python
def menorNumero(list):
    menor = list[0]
    idx = 0
    for i, num in enumerate(list):
        if num <= menor:
            menor = num
            idx = i
    return menor

def maiorNumero(list):
    maior = list[0]
    idx = 0
    for i, num in enumerate(list):
        if num >= maior:
            maior = num
            idx = i
    return maior

lista = []
while True:
    entrada = input('Insira um Número')
    if entrada == 'sair':
        break
    num = int(entrada)
    lista.append(num)


print(lista)
print(sum(lista)/len(lista))
print(menorNumero(lista))
print(maiorNumero(lista))
```
