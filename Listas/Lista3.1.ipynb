1. Soma dos elementos
Crie uma lista de números e calcule a soma de todos os elementos.

```python
nums = [1,2,3,4]

soma = 0
for num in nums:
    soma+=num
soma
```

2. Maior e menor
Encontre o maior e o menor valor de uma lista de inteiros.

```python
nums = [3,2,1,9,8,7]
menor = nums[0]
maior = nums[0]

for num in nums:
    if num < menor:
        menor = num 
    if num > maior: 
        maior = num 

print(menor)
print(maior)
```

3. Contar ocorrências
Conte quantas vezes um determinado elemento aparece em uma lista.

```python
nums = [1,2,3,3,4,5,5,5,6,7,8,8]
count = {}

for num in nums:
    count[num] = count.get(num, 0) + 1

count
```

4. Inverter lista
Inverta a ordem dos elementos de uma lista sem usar reverse() ou fatiamento
[::-1] (use um loop).

```python
nums = [5,4,3,2,1]

left = 0
right = len(nums)-1
while left < right:
    nums[left], nums[right] = nums[right], nums[left]
    left += 1
    right -= 1

nums
```

6. Intercalar listas
Dadas duas listas de mesmo tamanho, crie uma nova lista intercalando os
elementos (ex: [1,3,5] e [2,4,6] → [1,2,3,4,5,6]).

```python
nums1 = [1,3,5]
nums2 = [2,4,6]

nums = []

for n1, n2 in zip(nums1,nums2):
    nums.append(n1)
    nums.append(n2)
nums
```

7. Rotacionar lista
Rotacione uma lista para a direita por k posições (ex: [1,2,3,4] com k=2 →
[3,4,1,2]).

```python
nums = [1,2,3,4]

k = 2
n = len(nums)
res = [0] * n

for i, num in enumerate(nums):
    res[(i + k) % n] = nums[i]

res
```

8. Sublista contínua com maior soma
Encontre a sublista contínua (subarray) com a maior soma (algoritmo de
Kadane).

```python
nums = [4, -1, 2, -7, 3, 4, -2, 5, -1]

soma_atual = nums[0]
soma_max = nums[0]
inicio = 0
inicio_melhor = 0
fim_melhor = 0

for i, num in enumerate(nums):
    if num > soma_atual + num:
        soma_atual = num
        inicio = i 
    else:
        soma_atual += num 

    if soma_atual > soma_max:
        soma_max = soma_atual 
        inicio_melhor = inicio
        fim_melhor = i

nums[inicio_melhor:fim_melhor+1]
```

9. Produto de todos os elementos, exceto o atual
Dada uma lista de inteiros, retorne uma nova lista onde cada posição i é o
produto de todos os elementos da original exceto nums[i], sem usar divisão.

```python
nums = [1,2,3,4]

left = 1
right = 1
res = [1] * len(nums)

# Produto da Esquerda
prefix = 1
for i in range(len(nums)):
    res[i] = prefix
    prefix *= nums[i]

# Produto da Direita
suffix = 1
for i in range(len(nums)-1, -1, -1):
    res[i] *= suffix 
    suffix *= nums[i]

res
```

10. Merge de listas ordenadas (estilo merge sort)
Receba duas listas ordenadas e retorne uma única lista ordenada sem usar
sort() ou sorted()

```python
nums1 = [1, 4, 7, 12, 20]
nums2 = [2, 3, 8, 15, 18, 25]
i,j = 0,0

res = []

while i < len(nums1) and j < len(nums2):
    if nums1[i] < nums2[j]:
        res.append(nums1[i])
        i+=1
    else:
        res.append(nums2[j])
        j+=1

while i < len(nums1):
    res.append(nums1[i])
    i+=1

while j < len(nums2):
    res.append(nums2[j])
    j+=1

res
```
