```python
from multiprocessing import Process  # importa a classe Process para criar processos independentes

def f(name):
    # Função que será executada por cada processo filho.
    # Recebe um argumento 'name' e imprime uma mensagem identificando o processo.
    print('Hello, sou', name)

if __name__ == '__main__':
    # Garante que o bloco abaixo só será executado no processo principal.
    procs = []  # cria uma lista para armazenar os objetos Process

    for i in range(4):
        # Loop para criar 4 objetos Process.
        # Cada processo executará a função f com o índice i convertido para string.
        p = Process(target=f, args=(str(i),))
        procs.append(p)  # adiciona o objeto Process à lista, ainda não iniciado

    for i in range(4):
        # Loop para iniciar todos os processos criados.
        # start() cria o processo filho e faz com que ele execute a função alvo em paralelo.
        procs[i].start()

    for i in range(4):
        # Loop para aguardar a conclusão de cada processo.
        # join() bloqueia o processo principal até que o processo filho termine.
        procs[i].join()
        # após o join, sabemos que o processo procs[i] já finalizou
```

Exemplo - Código de PI

$\pi = \frac{p}{d}$

Diametro ($d$)    = $2r$ \
Perimetro ($p$)   = $2\pi r$ \
Area        = $\pi r^2$

Matematicamente sabemos que: \
$
\int_0^1 \frac{4.0}{1+x^2}dx = \pi
$

Podemos aproximar essa integral como a soma de retangulos: \
$
\sum_{i=0}^n F(x_i)\Delta x \cong \pi
$ \
Onde cada retângulo tem largura $\Delta x$ e altura $F(x_i)$ no meio do intervalo $i$

Código Sequencial:


Tempo gasto: 12.65 segundos

```python
import time 

def pi_naive(start,end, step):
    print("Start:", str(start))
    print("End:", str(end))

    sum  = 0.0 
    for i in range(start,end):
        x = (i + 0.5) * step
        sum += 4.0 / (1.0 + x * x)
    return sum

if __name__ == '__main__':
    num_steps = 100_000_000 
    sums = 0.0 
    step = 1.0 / num_steps 
    tic = time.time()
    sums = pi_naive(0, num_steps, step)
    toc = time.time()
    pi = step * sums
    print("pi:", pi)
    print("Tempo gasto:", toc - tic, "segundos")
```

Código Paralelo

Tempo de execução: 5.0056 segundos

```python
from multiprocessing import Process

def pi_naive(start, end, step):
    soma = 0.0
    for i in range(start, end):
        x = (i + 0.5) * step
        soma += 4.0 / (1.0 + x * x)
    return soma

if __name__ == "__main__":
    num_steps = 1_000
    step = 1.0 / num_steps

    n_proc = 4
    chunk = num_steps // n_proc

    processos = []

    for i in range(n_proc):
        inicio = i * chunk
        fim = num_steps if i == n_proc - 1 else (i + 1) * chunk

        processos.append(
            Process(
                target=pi_naive,
                args=(inicio, fim, step)
            )
        )

    for p in processos:
        p.start()

    for p in processos:
        p.join()

    print("FIM")
```
