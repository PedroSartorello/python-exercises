Strings

1. Inversão de string. Peça ao usuário uma palavra e exiba ela invertida (sem usar
fatiamento [::-1]. Use um laço).
2. Contador de vogais. Dada uma frase, conte quantas vogais (a, e, i, o, u; maiúsculas e
minúsculas) ela possui.
3. Validador de palíndromo. Verifique se uma palavra ou frase (ignorando espaços e
pontuação) é um palíndromo.
4. Gerador de sigla. Receba uma frase (ex.: "Organização das Nações Unidas") e retorne
a sigla em maiúsculas ("ONU").
5. Formatação de CPF. Dado um CPF como "12345678909", exiba no formato
"123.456.789-09".
 

```python
# 1
def inverterString(palavra: str) -> str:
    palavraInvertida = ''
    for i in range(len(palavra)-1, -1, -1):
        palavraInvertida += palavra[i]
    return palavraInvertida

inverterString('Pedro')

# 2
def contarVogais(palavra: str) -> str:
    vogais = {} # vogal : numVogal
    palavra = palavra.lower()
    for letra in palavra:
        if letra in vogais:
            vogais[letra] += 1
        elif letra in 'aeiou':
            vogais[letra] = 1
    return vogais

contarVogais('Pedro')

# 3 
def validarPalindromo(palavra: str) -> bool:
    palavra = palavra.lower()
    if palavra == inverterString(palavra):
        return True
    return False

validarPalindromo('Pedro')
validarPalindromo('ARARA')

# 4 
def gerarSigla(frase: str) -> str:
    sigla = ''
    palavras = frase.split() 
    for palavra in palavras:
        if palavra[0].isupper():
            sigla += palavra[0]
    return sigla

gerarSigla('Organização das Nações Unidas')

# "123.456.789-09".
def formatarCPF(cpf: str) -> str:
    cpfFormatado = (
        cpf[0:3] + '.' +
        cpf[3:6] + '.' +
        cpf[6:9] + '-' +
        cpf[9:11]
    )
    return cpfFormatado


print(formatarCPF('12345678909'))
```

O programa só sai quando a opção for 5. Trate divisão por zero.


```python
# 6 
def classificarTriangulo(l1: int, l2: int, l3: int) -> str:
    if l1 > 0 and l2 >0 and l3 > 0:
        if (l1 >= l2 + l3 or l2 >= l1 + l3 or l3 >= l1 + l2): 
            return 'Não forma Triangulo'
        if l1 == l2 == l3:
            return 'Equilátero'
        elif l1 == l2 or l1 == l3 or l2 == l3:
            return 'Isósceles'
        return 'Escaleno'
    return 'Valores devem ser positivos'

classificarTriangulo(20,600,20)

# 7 
def tabuadaInterativa() -> None:
    while True:
        num = int(input('Insira um número: '))
        for i in range(1, 11):
            print(f'{num} x {i} = {num*i}')
        continuar = input('Deseja calcular outro número? (s/n): ').lower()
        if continuar != 's':
            break
tabuadaInterativa()

# 8
import random 
def adivinhar() -> None:
    guessNum = 0
    rdNum = random.randint(1, 100)
    while True:
        guessNum = int(input('Qual o número?'))
        if guessNum > rdNum:
            print('Palpite maior que o número escolhido')
        elif guessNum < rdNum:
            print('Palpite menor que o número escolhido')
        else:
            print(f'Acertou! O número é {rdNum}')
            break
adivinhar()

# 9 
def somaPares()-> int | str:
    somaPares = 0
    a, b = map(int, input('Insira a e b (a < b): ').split())
    if not a < b:
        return 'a deve ser menor que b'
    for i in range(a,b+1):
        if (i % 2) == 0:
            somaPares += i
    return somaPares
somaPares()

# 10 
def menuOperacoes() -> None:
    while True:
        entrada = int(input('1 - Somar, 2 - Subtrair, 3 - Multiplicar, 4 - Dividir, 5 - Sair'))
        if entrada == 1:
            a, b = map(int, input('Insira a e b: ').split())
            soma = a + b
            print(f'A soma é {soma}')
        elif entrada == 2:
            a, b = map(int, input('Insira a e b: ').split())
            subtracao = a - b
            print(f'A subtracao é {subtracao}')
        elif entrada == 3:
            a, b = map(int, input('Insira a e b: ').split())
            mul = a * b
            print(f'A multiplicacao é {mul}')
        elif entrada == 4:
            a, b = map(int, input('Insira a e b: ').split())
            if b == 0:
                print('Não existe divisão por zero.')
            else: 
                div = a / b
                print(f'A divisao é {div}')
        elif entrada == 5:
            print('Saindo do programa....')
            break
        else:
            print('Opção inválida')

menuOperacoes()
```
