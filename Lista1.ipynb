Exercício 1 – Calculadora de IMC
Crie um programa que solicite peso (kg) e altura (m) do usuário, calcule o Índice de
Massa Corporal (IMC = peso / altura²) e classifique segundo a tabela WHO:
< 18.5 (Abaixo do peso), 18.5–24.9 (Normal), 25–29.9 (Sobrepeso), ≥ 30 (Obesidade).

```python
peso, altura = map(float, input('Peso (kg) e Altura (m)? ').split())

IMC = peso / altura**2

if IMC < 18.5:
    print('Abaixo do Peso')

elif IMC >= 18.5 and IMC < 25:
    print('Normal')

elif IMC >= 25 and IMC < 30:
    print('Sobrepeso')

else:
    print('Obesidade')
```

Exercício 10 – Gerenciador de tarefas com arquivo JSON
Crie um programa que:
● Carregue tarefas de tarefas.json (lista de dicionários com titulo, concluida
booleano)
● Permita adicionar, marcar como concluída, remover e salvar
● Trate exceções de leitura/escrita do arquivo

```python
import json 
with open('tarefas.json', encoding='utf-8') as arquivo:
    tarefas = json.load(arquivo)

def mostrarTarefas(tarefas: json) -> None:
    # Retorna a lista de tarefas e seu indicador (index do array tarefas)
    for i, tarefa in enumerate(tarefas):
        print(f"ID: {i} -- Titulo: {tarefa['titulo']} -- Concluida: {tarefa['concluida']}")
    return None

while True:
    opcao = int(input('1 - Adicionar Tarefa \n 2 - Concluir Tarefa \n 3 - Remover Tarefa \n 4 - Salvar Tarefa \n 5 - Sair'))
    match opcao:
        case 1:
            titulo = input('Escreva o Titulo da Tarefa')
            tarefa = {
                'titulo'    : titulo,
                'concluida' : False
            }
            tarefas.append(tarefa)
            print('Tarefa Adicionada')
        case 2: 
            mostrarTarefas(tarefas)
            idx = int(input('Selecione o ID da tarefa a ser concluida'))
            tarefas[idx]['concluida'] = True
            print('Tarefa Concluida com Sucesso')
        case 3: 
            mostrarTarefas(tarefas)
            idx = int(input('Selecione o ID da tarefa a ser removida'))
            tarefas.pop(idx)
            print('Tarefa Removida com Sucesso') 
        case 4:
            with open('tarefas.json', 'w', encoding='utf-8') as arquivo:
                json.dump(tarefas, arquivo, ensure_ascii=False, indent=4)
        case 5:
            print('Saindo do Programa...') 
            break
        case 6:
            mostrarTarefas(tarefas)

print('==========FIM===========')
```

Exercício 11 – Estatísticas descritivas sem NumPy
Crie duas listas de 10.000 números aleatórios (use random.randint ou random.gauss).
Implemente funções que calculem sem usar bibliotecas estatísticas:
● Média
● Mediana
● Moda (se houver)
● Desvio padrão amostral
Compare os resultados com statistics.mean, statistics.median, etc.

```python
import random 
import statistics
lista1 = [random.randint(0,10) for _ in range(10_000)]
lista2 = [random.gauss(0,1) for _ in range(10_000)]

def media(lista: list) -> float:
    print(f'Resultado da biblioteca: {statistics.mean(lista)}')
    return print(f'Resultado da funçao: {sum(lista)/len(lista)}')

def mediana(lista: list) -> float:
    print(f'Resultado da biblioteca: {statistics.median(lista)}') 
    lista = sorted(lista)
    tam = len(lista)
    if tam % 2 == 0:
        n1, n2 = lista[tam // 2], lista[tam//2 - 1]
        medianaVar = (n1 + n2) / 2
    else:
        medianaVar = lista[tam // 2]
    return print(f'Resultado da funçao: {medianaVar}')

def moda(lista: list) -> float:
    print(f'Resultado da biblioteca: {statistics.mode(lista)}')
    numeros = {}
    for num in lista:
        if num in numeros:
            numeros[num] += 1
        else:
            numeros[num] = 1
    return print(f'Resultado da função: {max(numeros, key = numeros.get)}')

def desvioPadraoAmostral(lista: list) -> float:
    print(f'Resultado da biblioteca: {statistics.stdev(lista)}')
    valor_media = sum(lista)/len(lista)
    stdev = 0
    for valor in lista:
        stdev += (valor - valor_media)**2
    stdev = (stdev/(len(lista)-1))**0.5
    return print(f'Resultado da função: {stdev}')

media(lista1)
mediana(lista1)
moda(lista1)
desvioPadraoAmostral(lista1)
media(lista2)
mediana(lista2)
moda(lista2)
desvioPadraoAmostral(lista2)
```

Exercício 12 – Matriz transposta e multiplicação manual
Implemente funções que:
● Transponham uma matriz (lista de listas)
● Multipliquem duas matrizes (verificar compatibilidade de dimensões)
Teste com uma matriz 3x3 e outra 3x2.

```python
A = [[1,2,3],
     [4,5,6],
      [7,8,9]]

B = [[10, 11],
     [12,13],
     [14,15]]

def transposta(matriz: list) -> list:
    linhas = len(matriz)
    colunas = len(matriz[0])

    M = [[0 for _ in range(linhas)] for _ in range(colunas)] 
    for i in range(linhas):
        for j in range(colunas):
            M[j][i] = matriz[i][j]
    
    return M

def multiplicar(A: list, B: list) -> list:
    colA, linhaB = len(A[0]), len(B)    
    if not (colA==linhaB):
        print("Não Compativel")
        return None 
    
    linhasA, colB = len(A), len(B[0])
    C = [[0 for _ in range(colB)] for _ in range(linhasA)]
    
    for i in range(linhasA):
        for j in range(colB):
            soma = 0
            for k in range(colA):
                soma+= A[i][k] * B[k][j]
            C[i][j] = soma
    
    return C
     
transposta(A)
transposta(B)
multiplicar(A,B)
multiplicar(B,A) 
```

Exercício 13 – Simulação de Monte Carlo para estimar π
Use o método de Monte Carlo: gere N pontos aleatórios (x, y) no quadrado [-1, 1] × [-1, 1] 
e conte quantos caem dentro do círculo unitário (x² + y² ≤ 1). 
Estime π como 4 × (pontos_no_círculo / N). 
Faça para N = 1.000, 10.000, 100.000, 1.000.000 
e exiba a precisão crescente.

```python
import random
N = [1_000, 10_000, 100_000, 1_000_000]

for numPontos in N:
    
    x = [random.uniform(-1, 1) for _ in range(numPontos)]
    y = [random.uniform(-1, 1) for _ in range(numPontos)]

    pontos = list(zip(x,y))

    circulo = [par for par in pontos if par[0]**2 + par[1]**2 <= 1]

    pi = 4 * (len(circulo)/numPontos)
    print(pi)
```

Exercício 14 – Análise de log de servidor
Dado o arquivo acessos.log, calcule:
● Número total de requisições
● Top 5 IPs que mais acessaram
● Quantas requisições por hora (0-23)
● Porcentagem de status 200 (sucesso) vs 404 (erro)

# Exemplo de Log

54.36.149.41 - - [22/Jan/2019:03:56:14 +0330] "GET /filter/27|13%20%D9%85%DA%AF%D8%A7%D9%BE%DB%8C%DA%A9%D8%B3%D9%84,27|%DA%A9%D9%85%D8%AA%D8%B1%20%D8%A7%D8%B2%205%20%D9%85%DA%AF%D8%A7%D9%BE%DB%8C%DA%A9%D8%B3%D9%84,p53 HTTP/1.1" 200 30577 "-" "Mozilla/5.0 (compatible; AhrefsBot/6.1; +http://ahrefs.com/robot/)" "-"

## Regex
- Pegar o IP:       (\S+)
- Pegar o traço:    - -
- Pegar a data:     \[(.*?)\]
- Pegar o GET:      (.*?)       # Tipo de Requisição
- Pegar Código:     (\d+) (\d+) # Status e Tamanho
- Pegar o traço:    -
- Pegar o referer e user agent:  "(.*?)" "(.*?)"



```python
import re

padrao = r'(\S+) - - \[(.*?)\] "(.*?)" (\d+) (\d+) "(.*?)" "(.*?)"'
dados = []

with open('acessos.log.txt', 'r', encoding='utf-8') as f:
    for linha in f:
        match = re.match(padrao, linha)
        
        if match:
            dados.append({
                'ip'        :  match.group(1),
                'data'      :  match.group(2),
                'request'   :  match.group(3),
                'status'    :  int(match.group(4)),
                'size'      :  int(match.group(5)),
                'referer'   :  match.group(6),
                'agent'     :  match.group(7)
            })

# Número total de requisições
numRequisicoes = len(dados)
print(f"Número total de requisições: {numRequisicoes}")

# Top 5 IPs que mais acessaram
ipDict = {}   # IP - Contagem
for dado in dados:
    if dado['ip'] in ipDict:
        ipDict[dado['ip']] += 1
    else:
        ipDict[dado['ip']] = 1
top5 = sorted(ipDict.items(),key = lambda x: x[1] ,reverse=True)[:5]
print(f"Top 5 IPs que mais acessaram: {top5}")

# Quantas requisições por segundo
segundos = {}
for dado in dados:
    timestamp =  dado['data'].split(' ')[0]
    if timestamp in segundos:
        segundos[timestamp] += 1
    else:
        segundos[timestamp] = 1
Media_reqPorSegundo = len(segundos) / len(dados)
print(f"Quantas requisições por segundo: {segundos}")

# Porcentagem de status 200 (sucesso) vs 404 (erro)
cont200 = 0
cont404 = 0
for dado in dados:
    status = dado['status']
    if status == 200:
        cont200 += 1
    if status == 404:
        cont404 += 1

status200 = 100 * (cont200/len(dados))
status404 = 100 * (cont404/len(dados))

print(f'200 -> {status200:.3f}%')
print(f'404 -> {status404:.3f}%')
```
